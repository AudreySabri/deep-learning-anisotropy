```python
def plot_mlp_predictions(predictions, ground_truth, model_dir, c=None, c_label = None, feature_names=None, filename="predictions.png"):
    """Plots the predictions and saves the plot to model_dir
    
    Args:
        predictions: (numpy array) predicted values
        scores: (numpy array) scores for the predictions
        ground_truth: (numpy array) true values
        model_dir: (string) directory where the plot is to be saved
        feature_names: (list) names of the target features
        filename: (string) base filename for saved plots
    """
    import matplotlib.pyplot as plt
    import numpy as np
    import os
    
    # Define default feature names if not provided
    if feature_names is None:
        feature_names = [f'hill_coeff_{i+1}' for i in range(6)] + \
                        [f'rotation_{comp}' for comp in ['11', '22', '33', '23', '13', '12', 
                                                          '32', '31', '21']]  # Assuming Voigt notation
    
    # Ensure we have enough names
    n_features = predictions.shape[1] if len(predictions.shape) > 1 else 1
    if n_features > len(feature_names):
        feature_names = feature_names + [f'feature_{i+1}' for i in range(len(feature_names), n_features)]
    
    # Calculate metrics
    mae = np.mean(np.abs(predictions - ground_truth), axis=0)
    mse = np.mean((predictions - ground_truth)**2, axis=0)
    r2 = 1 - np.sum((ground_truth - predictions)**2, axis=0) / np.sum((ground_truth - np.mean(ground_truth, axis=0))**2, axis=0)
    
    # Create subplots
    n_cols = 3
    n_rows = (n_features + n_cols - 1) // n_cols
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(7*n_cols, 7*n_rows))
    axes = axes.flatten() if n_features > 1 else [axes]
    
    for idx in range(n_features):
        ax = axes[idx]
        
        # Extract data for this feature
        if n_features > 1:
            pred = predictions[:, idx]
            true = ground_truth[:, idx]
        #    score = scores[:, idx] if len(scores.shape) > 1 else scores
        else:
            pred = predictions
            true = ground_truth
      #      score = scores
        
        # Create scatter plot with error bars
        if c is not None:
            scatter = ax.scatter(true, pred, alpha=0.6, c=c,
                            s=30, edgecolors='black', linewidth=0.5, rasterized=True)        
        else:
            scatter = ax.scatter(true, pred, alpha=0.6, 
                            s=30, edgecolors='black', linewidth=0.5, rasterized=True)

        #ax.errorbar(true, pred, yerr=score, fmt='none', alpha=0.2, color='gray')
        
        # Add perfect prediction line
        min_val = min(true.min(), pred.min())
        max_val = max(true.max(), pred.max())
        ax.plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.8, 
                linewidth=2, label='Perfect prediction')
        
        # Add metrics to plot
        ax.text(0.05, 0.95, f'MAE: {mae[idx]:.3f}\nR²: {r2[idx]:.3f}', 
                transform=ax.transAxes, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5), fontsize='xx-large')

        ax.scatter(
        mlp_carp_true[:, idx],
        mlp_carp_pred[:, idx],
        s=50,
        marker='s',
        facecolors='none',
        edgecolors='orange',
        linewidths=2,
        zorder=10,
        alpha = 0.7,
        rasterized=True,
        label=f'Carpathes DB (MAE={mlp_carp_mae[idx]:.2f})'
    )
    
        # Highlight POLY point(s)
        ax.scatter(
            mlp_poly_true[:, idx],
            mlp_poly_pred[:, idx],
            s=50,
            marker='D',
            facecolors='none',
            edgecolors='red',
            linewidths=2,
            zorder=10,
            alpha = 0.7,
            rasterized=True,
            label=f'Polynesia DB (MAE={mlp_poly_mae[idx]:.2f})'
        )

        
        # Labels and title
        ax.set_xlabel('Ground Truth', fontsize='x-large')
        ax.set_ylabel('Predictions', fontsize='x-large')
        ax.set_title(f'{feature_names[idx]}', fontsize='xx-large')
        ax.legend(loc='lower right')
        ax.grid(True, alpha=0.3)
        ax.tick_params(axis='both', labelsize=15)
        ax.set_xlim(min_val, max_val)
        ax.set_ylim(min_val, max_val)
        
        # Set equal aspect ratio
        ax.set_aspect('equal', adjustable='box')

        if c is not None:
            fig.colorbar(scatter, ax=ax, label = c_label)
    
    # Hide empty subplots
    for idx in range(n_features, len(axes)):
        axes[idx].set_visible(False)
    
  #  plt.tight_layout()
    
    # Save the plot
    save_path = os.path.join(model_dir, filename)
    plt.savefig(save_path,bbox_inches='tight', format = 'svg')

def plot_bnn_predictions(predictions, ground_truth, lower_bound, upper_bound, model_dir, c=None, c_label = None, feature_names=None, filename="predictions.png"):
    """Plots the predictions and saves the plot to model_dir
    
    Args:
        predictions: (numpy array) predicted values
        scores: (numpy array) scores for the predictions
        ground_truth: (numpy array) true values
        model_dir: (string) directory where the plot is to be saved
        feature_names: (list) names of the target features
        filename: (string) base filename for saved plots
    """
    import matplotlib.pyplot as plt
    import numpy as np
    import os

    import matplotlib as mpl
    mpl.rcParams['agg.path.chunksize'] = 1000
    
    # Define default feature names if not provided
    if feature_names is None:
        feature_names = [f'hill_coeff_{i+1}' for i in range(6)] + \
                        [f'rotation_{comp}' for comp in ['11', '22', '33', '23', '13', '12', 
                                                          '32', '31', '21']]  # Assuming Voigt notation
    
    # Ensure we have enough names
    n_features = predictions.shape[1] if len(predictions.shape) > 1 else 1
    if n_features > len(feature_names):
        feature_names = feature_names + [f'feature_{i+1}' for i in range(len(feature_names), n_features)]
    
    # Calculate metrics
    mae = np.mean(np.abs(predictions - ground_truth), axis=0)
    mse = np.mean((predictions - ground_truth)**2, axis=0)
    r2 = 1 - np.sum((ground_truth - predictions)**2, axis=0) / np.sum((ground_truth - np.mean(ground_truth, axis=0))**2, axis=0)
    
    # Create subplots
    n_cols = 3
    n_rows = (n_features + n_cols - 1) // n_cols
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(7*n_cols, 7*n_rows))
    axes = axes.flatten() if n_features > 1 else [axes]
    
    for idx in range(n_features):
        ax = axes[idx]
        
        # Extract data for this feature
        if n_features > 1:
            pred = predictions[:, idx]
            true = ground_truth[:, idx]
            lower_bound_feature = lower_bound[:, idx]
            upper_bound_feature = upper_bound[:, idx]
        #    score = scores[:, idx] if len(scores.shape) > 1 else scores
        else:
            pred = predictions
            true = ground_truth
            lower_bound_feature = lower_bound
            upper_bound_feature = upper_bound
      #      score = scores
        
        # Create scatter plot with error bars
        if c is not None:
            scatter = ax.scatter(true, pred, alpha=0.6, c=c,
                            s=30, edgecolors='black', linewidth=0.5, rasterized=True)        
        else:
            scatter = ax.scatter(true, pred, alpha=0.6, 
                            s=30, edgecolors='black', linewidth=0.5, rasterized=True)
        #ax.errorbar(true, pred, yerr=score, fmt='none', alpha=0.2, color='gray')
        sort_idx = np.argsort(true)

        true_sorted = true[sort_idx]
        lower_sorted = lower_bound_feature[sort_idx]
        upper_sorted = upper_bound_feature[sort_idx]

        fill = ax.fill_between(
            true_sorted,
            lower_sorted,
            upper_sorted,
            color='green',
            alpha=0.1,
            label='Uncertainty interval',
            rasterized=True
            )

        ax.scatter(
        bnn_carp_true[:, idx],
        bnn_hill_carp[:, idx],
        s=50,
        marker='s',
        facecolors='white',
        edgecolors='orange',
        linewidths=2,
        zorder=10,
        alpha = 0.7,
        rasterized=True,
        label=f'Carpathes DB (MAE={carp_mae[idx]:.2f})'
    )
    
        # Highlight POLY point(s)
        ax.scatter(
            bnn_poly_true[:, idx],
            bnn_hill_poly[:, idx],
            s=50,
            marker='D',
            facecolors='white',
            edgecolors='red',
            linewidths=2,
            zorder=10,
            alpha = 0.7,
            rasterized=True,
            label=f'Polynesia DB (MAE={poly_mae[idx]:.2f})'
        )
        
        # Add perfect prediction line
        min_val = min(true.min(), pred.min())
        max_val = max(true.max(), pred.max())
        ax.plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.8, 
                linewidth=2, label='Perfect prediction')
        
        # Add metrics to plot
        ax.text(0.05, 0.95, f'MAE: {mae[idx]:.3f}\nR²: {r2[idx]:.3f}', 
                transform=ax.transAxes, verticalalignment='top',
                bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5), fontsize='xx-large')

        
        # Labels and title
        ax.set_xlabel('Ground Truth', fontsize='x-large')
        ax.set_ylabel('Predictions', fontsize='x-large')
        ax.set_title(f'{feature_names[idx]}', fontsize='xx-large')
        ax.legend(loc='lower right')
        ax.grid(True, alpha=0.3)
        ax.tick_params(axis='both', labelsize=15)
        ax.set_xlim(min_val, max_val)
        ax.set_ylim(min_val, max_val)
        
        # Set equal aspect ratio
        ax.set_aspect('equal', adjustable='box')
        if c is not None:
            fig.colorbar(scatter, ax=ax, label = c_label)
    # Hide empty subplots
    for idx in range(n_features, len(axes)):
        axes[idx].set_visible(False)
        
    # Save the plot
    save_path = os.path.join(model_dir, filename)
    plt.savefig(save_path,bbox_inches='tight', format = 'svg')
 #   plt.close()

def load_data(csv_path, model='mlp'):
    """Load rotation matrices and MAE data from CSV file.
    
    Extracts columns starting with 'ground_truth' as ground truth,
    columns starting with 'obs_mean' as predictions, and MAE data.
    """
    df = pd.read_csv(csv_path)
    
    # Extract ground truth columns
    ground_truth_cols = sorted([col for col in df.columns if col.startswith('ground_truth')])
    # Extract prediction columns
    if model == 'mlp':
        pred_cols = sorted([col for col in df.columns if col.startswith('prediction')])
    if model == 'bnn':
        pred_cols = sorted([col for col in df.columns if col.startswith(('mean', '5', '95', 'std'))])
    if model == 'mlp':
        return df[ground_truth_cols].values, df[pred_cols].values
    if model == 'bnn':
        return df[ground_truth_cols].values, df[pred_cols]
```

```python
def load_column(csv_path, column='strain', model='mlp'):
    import torch
    df = pd.read_csv(csv_path, usecols=[column], dtype=np.float32)
    df_values = df.values
    dataset_size = len(df)
    generator = torch.Generator().manual_seed(42)
    indices = torch.randperm(dataset_size, generator=generator)
    if model == 'mlp': 
        test_size = int(dataset_size*0.2)
        test_indices = indices[-test_size:]
        test_values = df_values[test_indices]
    if model == 'bnn': 
        test_size = int(dataset_size*0.2)
        test_indices = indices[-test_size:]
        test_values = df_values[test_indices]
    return test_values
```

```python
from matplotlib import cm
from matplotlib.colors import Normalize
from scipy.stats import binned_statistic


def colored_hist_by_value(
    ax,
    values,
    colors_by=None,
    bins=60,
    cmap="viridis",
    density=False,      # if True → probability density instead of %
    as_percent=True,    # if True → percentage instead of raw counts
    edgecolor="k"
):
    values = np.asarray(values).reshape(-1)

    # -----------------------------
    # 1. Histogram (counts)
    # -----------------------------
    counts, bin_edges = np.histogram(values, bins=bins)

    bin_width = bin_edges[1] - bin_edges[0]
    bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])

    # Convert to frequency / percentage
    if density:
        heights = counts / (counts.sum() * bin_width)
        ylabel = "Density"
    elif as_percent:
        heights = 100.0 * counts / counts.sum()
        ylabel = "Percentage (%)"
    else:
        heights = counts
        ylabel = "Count"

    # -----------------------------
    # 2. Compute per-bin mean of colors_by (IMPORTANT FIX)
    # -----------------------------
    if colors_by is not None:
        colors_by = np.asarray(colors_by).reshape(-1)

        if len(colors_by) != len(values):
            raise ValueError(
                f"colors_by and values must have same length: "
                f"{len(colors_by)} vs {len(values)}"
            )

        per_bin_stats, _, _ = binned_statistic(
            values,
            colors_by,
            statistic="mean",
            bins=bin_edges
        )

        # -----------------------------
        # 3. Robust normalization (avoids outlier color collapse)
        # -----------------------------
        vmin = np.nanpercentile(per_bin_stats, 5)
        vmax = np.nanpercentile(per_bin_stats, 95)

        if vmin == vmax:
            vmin, vmax = np.nanmin(per_bin_stats), np.nanmax(per_bin_stats)

        norm = Normalize(vmin=vmin, vmax=vmax)
        cmap_obj = cm.get_cmap(cmap)

        bar_colors = [
            cmap_obj(norm(v)) if not np.isnan(v) else (0.85, 0.85, 0.85, 1.0)
            for v in per_bin_stats
        ]

        sm = cm.ScalarMappable(norm=norm, cmap=cmap_obj)
        sm.set_array([])

    else:
        bar_colors = "blue"
        sm = None

    # -----------------------------
    # 4. Plot
    # -----------------------------
    ax.bar(
        bin_centers,
        heights,
        width=bin_width * 0.9,
        color=bar_colors,
        edgecolor=edgecolor
    )

    ax.set_ylabel(ylabel)

    return sm

def plot_errors(t1_file, strain, out_file, out_prefix='/home/sabria/scratch_sabria/deep-learning-anisotropy/plots/bnn_hill/'):
    t1 = np.load(t1_file)
    t1 = t1.T
  #  t2_vals, t2_omegas = t2
  #  t2_vals = np.vstack(t2_vals)

    stress_names = ['traction xx', 'traction yy', 'traction zz', 'cisaillement yz', 'cisaillement zx', 'cisaillement xy']

  #  strain = load_strain('/home/sabria/scratch_sabria/vpsc-hill/data/aug_db.csv')
   # J = pd.read_csv('/home/sabria/scratch_sabria/vpsc-hill/data/cf_db.csv', usecols=['J'], dtype=np.float32)['J'].values

    # Plot Test1 2x3
    fig1, axes1 = plt.subplots(2, 3, figsize=(20, 10))
    axes1 = axes1.ravel()
  #  sm1 = None
    for idx in range(6):
        ax = axes1[idx]
        vals = t1[:, idx]
        sm1 = colored_hist_by_value(ax, vals, strain, bins=20)
        t1_stats = [(np.mean(t1[:, i]), np.std(t1[:, i]), np.median(t1[:, i])) for i in range(6)]
        textstr = '\n'.join((
            r'$\mu=%.2f$' % (t1_stats[idx][0], ),
            r'$\mathrm{median}=%.2f$' % (t1_stats[idx][2], ),
            r'$\sigma=%.2f$' % (t1_stats[idx][1], )))
        props = dict(boxstyle='round', facecolor='wheat', alpha=0.5)
        ax.text(0.5, 0.95, textstr, transform=ax.transAxes, fontsize=14,
            verticalalignment='top', bbox=props)
        ax.set_title(stress_names[idx])
        ax.set_xlabel('Error due to Hill predictions (%)')
        ax.set_xlim(0, 100)
        ax.set_xticks(np.arange(0, 101, 10))
    fig1.colorbar(sm1, ax=axes1.tolist(), orientation='vertical', label='Mean Strain per Bin')
    fig1.savefig(out_prefix + out_file, bbox_inches='tight', format='svg')
```

```python
import os
import json

def parse_training_log(log_file):
    train_losses = []

    pattern = re.compile(r'loss:\s*(-?\d+(?:\.\d+)?)')

    with open(log_file, 'r', encoding='utf-8', errors='ignore') as f:
        for line in f:
            match = pattern.search(line)
            if match:
                train_losses.append(float(match.group(1)))

    return train_losses

def plot_results(model_dir, out_prefix, out_file):
    """Plot training results from a model directory"""
    
    # Parse training log
    log_file = os.path.join(model_dir, 'bnn_main.log')
    train_losses= parse_training_log(log_file)
        
    # Create plot
    fig, axes = plt.subplots(1, 1, figsize=(7, 5))
    
    # Plot loss
    axes.plot(np.arange(1, 150, 5), train_losses, 
                 'b-o', label='Training Loss', linewidth=2)
    
    axes.set_xlabel('Epoch')
    axes.set_ylabel('Loss')
    axes.set_title('Training Loss')
    axes.legend()
    axes.grid(True, alpha=0.3)
    

    plt.tight_layout()
    plt.savefig(os.path.join(out_prefix, out_file), bbox_inches='tight', format='svg')
```

```python
import os
import numpy as np
import math
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

def correlation_analysis(
    csv_path,
    feature_pattern=r"^c\d+_out$",
    hill_pattern=r"^hill\d+_out$",
    save_prefix=None,
    aug = False
):

    df_sample = pd.read_csv(csv_path, nrows=1)

    feature_cols = list(df_sample.filter(regex=feature_pattern).columns)
    hill_cols = list(df_sample.filter(regex=hill_pattern).columns)
    name_col = "name"

    usecols = [name_col] + feature_cols + hill_cols

    dtype_map = {col: np.float32 for col in feature_cols + hill_cols}

    df = pd.read_csv(csv_path, usecols=usecols, dtype=dtype_map)

    if aug==False:
        df = df[~df[name_col].str.contains("ROT", na=False)]
        hill_out = df[hill_cols].to_numpy(copy=False).reshape(len(df),6)
        C_out = df[['c11_out', 'c12_out', 'c13_out',
                    'c22_out', 'c23_out',
                    'c33_out',
                    'c44_out',
                    'c55_out',
                    'c66_out']].to_numpy().reshape(len(df),9)
    else:
        hill_out = df[hill_cols].to_numpy(copy=False).reshape(len(df),6)
        C_out = df[['c11_out', 'c12_out', 'c13_out', 'c14_out','c15_out','c16_out',
                    'c22_out','c23_out','c24_out','c25_out','c26_out',
                    'c33_out','c34_out','c35_out','c36_out',
                    'c44_out','c45_out','c46_out',
                    'c55_out','c56_out',
                    'c66_out']].to_numpy().reshape(len(df),21)    
    if not aug:
        C_cols = [
            "C11", "C22", "C33",
            "C44", "C55", "C66",
            "C12", "C13", "C23"
        ]
    else:
        C_cols = [
            "C11","C22","C33","C44","C55","C66",
            "C12","C13","C14","C15","C16",
            "C23","C24","C25","C26",
            "C34","C35","C36",
            "C45","C46","C56"
        ]

    hill_labels = ["F", "G", "H", "L", "M", "N"]

    corr = np.zeros((len(hill_cols), len(C_cols)))

    for j in range(len(hill_cols)):
        for k in range(len(C_cols)):
            corr[j, k] = np.corrcoef(
                hill_out[:, j],
                C_out[:, k]
            )[0, 1]

    plt.figure(figsize=(1.2 * len(C_cols), 5))

    sns.heatmap(
        corr,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        linewidths=0.5,
        xticklabels=C_cols,
        yticklabels=hill_labels,
        cbar_kws={"label": "Pearson correlation"},
    )

    plt.xlabel("Elastic constants")
    plt.ylabel("Hill coefficients")
    plt.title(
        f"Correlation matrix ({'with' if aug else 'without'} augmentation)"
    )
```

```python
csv_files = [
        "aug_db_ac.csv",
        "aug_db_ae.csv",
    ]
csv_dir = '/home/sabria/scratch_sabria/vpsc-hill/tests/vpsc-HillCoeffs/COMPRY'

out_dir = "/home/sabria/scratch_sabria/vpsc-hill/plots/paper_plots/db_plots"

for csv_path in csv_files:
    df_path = os.path.join(csv_dir,csv_path)
    base_name = os.path.splitext(os.path.basename(csv_path))[0]

    no_aug_prefix = os.path.join(out_dir, f"no_aug_{base_name}")
    w_aug_prefix = os.path.join(out_dir, f"aug_{base_name}")
    
    correlation_analysis(
            csv_path=df_path,
            feature_pattern=r"^c\d+_out$",
            hill_pattern=r"^hill\d+_out$",
            save_prefix=no_aug_prefix,
            aug = False
        )
        
    correlation_analysis(
            csv_path=df_path,
            feature_pattern=r"^c\d+_out$",
            hill_pattern=r"^hill\d+_out$",
            save_prefix=w_aug_prefix,
            aug = True
        )
```
